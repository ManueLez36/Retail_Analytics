```python
# Asegurarnos de que el esquema Gold exista
spark.sql("CREATE DATABASE IF NOT EXISTS retail.03_gold")

# ==========================================
# 1. Crear Dimensión Clientes
# ==========================================
df_clientes_silver = spark.table("retail.02_silver.clientes")

dim_clientes = df_clientes_silver.select(
    "ID_Cliente",
    "Edad",
    "Sexo",
    "EstCivil",
    "Segmento_RFM"
).dropDuplicates(["ID_Cliente"])

dim_clientes.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail.03_gold.dim_clientes")



print("¡Modelo dimensional Gold generado con éxito!")
```

```python
# ==========================================
# 2. Crear Tabla de Hechos de Ventas
# ==========================================
df_ventas_silver = spark.table("retail.02_silver.ventas")

fact_ventas = df_ventas_silver.select(
    "CodTicket",
    "Dia",
    "ID_Cliente",
    "CodMaterial",
    "SKU",
    "Producto",
    "Descripción",
    "Tipo",
    "VTABS",
    "Margen",
    "VTAQ",
    "IGV"
)

fact_ventas.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail.03_gold.fact_ventas")

print("¡Modelo hechos Gold generado con éxito!")
```
