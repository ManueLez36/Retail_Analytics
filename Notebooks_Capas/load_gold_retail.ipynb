```python
# ==========================================
# 1. Cargar y poblar Dimensión Clientes en Gold
# ==========================================
df_clientes_silver = spark.table("retail.02_silver.clientes")

dim_clientes = df_clientes_silver.select(
    "ID_Cliente",
    "Edad",
    "Sexo",
    "EstCivil",
    "Segmento_RFM"
).dropDuplicates(["ID_Cliente"])

dim_clientes.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail.03_gold.dim_clientes")

print("¡Dimensión clientes cargada con éxito en Gold!")
```

```python
%sql
select * from retail.03_gold.dim_clientes
```

```python
# ==========================================
# 2. Cargar y poblar Hechos de Ventas en Gold
# ==========================================
df_ventas_silver = spark.table("retail.02_silver.ventas")

fact_ventas = df_ventas_silver.select(
    "CodTicket",
    "ID_Cliente",
    "CodMaterial",
    "SKU",
    "Producto",
    "Descripción", # Con su tilde correcta
    "Tipo",
    "VTABS",
    "Margen",
    "VTAQ",
    "IGV"
)

fact_ventas.write \
    .mode("overwrite") \
    .option("overwriteSchema", "true") \
    .saveAsTable("retail.03_gold.fact_ventas")

print("¡Tabla de hechos de ventas cargada con éxito en Gold!")
```

```python
%sql
select * from retail.03_gold.fact_ventas
limit(5)
```
