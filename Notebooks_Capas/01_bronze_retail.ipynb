```python
ventas_df = spark.read.csv("/Volumes/retail/dataset/landing/Data_Ventas.csv", sep=';', header=True, inferSchema=True)
display(ventas_df.limit(5))
```

```python
%sql
CREATE TABLE IF NOT EXISTS retail.01_bronze.ventas_df(
    Mes int,
    Dia int,
    CodTicket string,
    ID_Cliente int,
    CodMaterial int, 
    Marca string,
    Cod_N1 int,
    Desc_N1 string,
    N2 string,
    N3 string,
    VTABS double,
    Margen double,
    VTAQ double,
    IGV double
)
USING delta
    
```

```python
ventas_df.printSchema()
```

```python
%sql
select * from retail.01_bronze.ventas_df
limit 5
```

```python
clientes_df = spark.read.csv("/Volumes/retail/dataset/landing/Clientes.csv", sep =';', header=True, inferSchema=True)
display(clientes_df.limit(5))
```
